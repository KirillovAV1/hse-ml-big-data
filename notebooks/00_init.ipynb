```python
import os
import psycopg2
import numpy as np
import pandas as pd
import sqlalchemy as sql
```

### Считывание переменных окружения + формирование строки подключения

```python
class Settings():
    DB_NAME = os.getenv("DB_NAME")
    DB_HOST = os.getenv("DB_HOST")
    DB_PORT = os.getenv("DB_PORT")
    DB_USER = os.getenv("DB_USER")
    DB_PASSWORD = os.getenv("DB_PASSWORD")

    S3_URL = os.getenv("S3_URL")
    S3_USER = os.getenv("S3_USER")
    S3_PASSWORD = os.getenv("S3_PASSWORD")

    @property
    def DB_URL(self):
        return f"postgresql+psycopg2://{self.DB_USER}:{self.DB_PASSWORD}@{self.DB_HOST}:{self.DB_PORT}/{self.DB_NAME}"

settings = Settings()
```

### Создание движка и вывод списка доступных для работы таблиц в БД

```python
engine = sql.create_engine(settings.DB_URL)
pd.read_sql("SELECT table_name FROM information_schema.tables WHERE table_schema = 'public';", engine)
```
